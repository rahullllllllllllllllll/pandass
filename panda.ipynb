PANDAS ,, IT HAS THE COLUMNS CALLED AS SERIES RESPECTIVELY ONE BY ONE . 

```python
import pandas as pd
import numpy as np
# series from list
country=['rwanda','russia','romania']
pd.Series(country)
```

```python
# u can also give your own index 
rank=[1,7,9,67]
countr=['india','indonseia','ireland','iran']
pd.Series(rank,index=countr,name="the countries wit i")
```

```python
# series from dictionary
dict={'india':1,'indonseia':7,'ireland':9,'iran':67}
nuu=pd.Series(dict,name="same")# in it name will give a attribute to it , wheras countr is like a reference variable
```

```python
# series attributes
nuu.size
nuu.dtype
nuu.is_unique
nuu.index#will give index object with indexes in it , for intger index it give rangeindex object like not the int theitr range
nuu.values
```
